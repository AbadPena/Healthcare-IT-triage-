````python
devices = [
    {"name": "ER-WS-12", "type": "Windows workstation", "os": "Windows 11", "days_since_patch": 12, "encrypted": True},
    {"name": "NURSE-TAB-04", "type": "Tablet", "os": "Android", "days_since_patch": 47, "encrypted": True},
    {"name": "PATIENT-TAB-01", "type": "Tablet", "os": "Android", "days_since_patch": 18, "encrypted": True},
    {"name": "LAB-WS-03", "type": "Windows workstation", "os": "Windows 10", "days_since_patch": 95, "encrypted": False},
    {"name": "IMAGING-01", "type": "Imaging console", "os": "Vendor appliance", "days_since_patch": 160, "encrypted": True}
]

print(f"Loaded {len(devices)} fictional devices.")
````

````python
# I learned that even the smallest mistake can cause a syntax error, if I delete a quotation mark than the code wll break.
````

````python
for device in devices:
    print(f"Device: {device['name']:<13} | OS: {device['os']:<17} | {device['days_since_patch']} days unpatched")
````

````python
needs_attention = []

for device in devices:
    if device["days_since_patch"] > 60 or device["encrypted"] == False:
        needs_attention.append(device)

print(f"Devices needing review: {len(needs_attention)}")
for device in needs_attention:
    print('-', device["name"])
````
